### ChromaDB

```python
import chromadb
chroma_client = chromadb.Client()
```

```python
collection_name = "test_collection"
collection = chroma_client.get_or_create_collection(collection_name) # Collection for storing embeddings,documents,metadata etc
```

```python
collection.add(
    ids=["id1", "id2","id3"],
    documents=[
        "Hello world",
        "How are you today?",
        "GoodBye,see you later!"
    ]
)
```

```python
query_text = "Hello world"

results = collection.query(
    query_texts = [query_text],
    n_results=3,
)
print(results)
```
