```python
from sentence_transformers import SentenceTransformer 
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
# model
model = SentenceTransformer("all-MiniLM-L6-v2")

# sample text
text = """
langchain is a framework for building applications with LLMS.
langchain provides modular obstractions to combine llms with tools like openai and pinecone.
langchain is a popular framework.
you can create chains, agents, memory and retrievers.
The eiffel tower is located in paris.
France is a popular tourist destination which capital is paris.""" 

# Step 1: Split into sentences
sentences = [s.strip() for s in text.split("\n") if s.strip()]

# Step 2: Embed each sentences
embeddings = model.encode(sentences)

# Step 3: Initialize parameters 
threshold = 0.5
chunks = []
current_chunk = [sentences[0]]

# Step 4: Semantic grouping based on threshold
for i in range(1, len(sentences)):
    sim = cosine_similarity(
        [embeddings[i-1]],
        [embeddings[i]]
    )[0][0]
    if sim >= threshold:
        current_chunk.append(sentences[i])
    else:
        chunks.append(" ".join(current_chunk))
        current_chunk = [sentences[i]]
    
# Append the last chunk
chunks.append(" ".join(current_chunk))

# Output the chunks
print("Semantic chunks: ")
for idx,chunk in enumerate (chunks):
    print(f"\nChunk {idx+1}:\n{chunk}")
```
